```python
import pandas as pd
import re
```

```python
df = pd.read_csv(r"C:\Users\91882\Desktop\AI-Based Email Spam Detection\data\email_text.csv")
df['text'].iloc[0]
```

## Converting to lowercase

```python
df['clean_text'] = df['text'].str.lower()
```

```python
df[['text','clean_text']].head()
```

## Removing HTML tags :

```python
def remove_html(text):
    return re.sub(r'<.*?>', '', text)
```

```python
df['clean_text'] = df['clean_text'].apply(remove_html)
```

```python
df[['text', 'clean_text']].head()
```

```python
print(df.loc[0,'clean_text'])
```

```python
print(df.loc[0,'text'])
```

## Removing URLS :

```python
def remove_urls(text):
    return re.sub(r'http\S+|www\S+', '', text)
```

```python
df['clean_text'] = df['clean_text'].apply(remove_urls)
```

```python
print(df.loc[0,'clean_text'])
```

## Removing Special Characters :

```python
def remove_special_chars(text):
    return re.sub(r'[^a-zA-Z0-9\s]', '', text)
```

```python
df['clean_text'] = df['clean_text'].apply(remove_special_chars)
```

```python
print(df.loc[0,'clean_text'])
```

## Removing Extra Spaces :

```python
def remove_extra_spaces(text):
    return re.sub(r'\s+', ' ', text).strip()
```

```python
df['clean_text'] = df['clean_text'].apply(remove_extra_spaces)
```

```python
df['clean_text'].iloc[0]
```

## Removing escape artifacts :

```python
def remove_dataset_artifacts(text):
    text = re.sub(r'escapenumber\w*', '', text)
    text = re.sub(r'escapelong', '', text)
    return text
```

```python
df['clean_text'] = df['clean_text'].apply(remove_dataset_artifacts)
```

### Again remove extra spaces caused by previous operation

```python
df['clean_text'] = df['clean_text'].apply(remove_extra_spaces)
```

```python
df['clean_text'].iloc[0]
```

## Stopword Removal : (eg : is , the , etc ...)

```python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

def remove_stopwords(text):
    words = text.split()
    words = [word for word in words if word not in ENGLISH_STOP_WORDS]
    return ' '.join(words)
```

```python
df['clean_text'] = df['clean_text'].apply(remove_stopwords)
```

```python
df['clean_text'].iloc[0]
```
